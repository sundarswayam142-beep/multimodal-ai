```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('your_dataset.csv')

# 1. Initial inspection
df.info()
df.describe()
df.head()

# 2. Check missing values
df.isnull().sum()

# 3. Handle missing values
df['column_name'] = df['column_name'].fillna(df['column_name'].mean())  # numeric
df['category_col'] = df['category_col'].fillna(df['category_col'].mode()[0])  # categorical
df.dropna(subset=['critical_column'], inplace=True)  # drop rows if critical col is null

# 4. Check and remove duplicates
df.duplicated().sum()
df.drop_duplicates(inplace=True)

# 5. Fix data types
df['date_column'] = pd.to_datetime(df['date_column'])
df['numeric_col'] = pd.to_numeric(df['numeric_col'], errors='coerce')

# 6. Handle outliers (IQR method)
Q1 = df['numeric_col'].quantile(0.25)
Q3 = df['numeric_col'].quantile(0.75)
IQR = Q3 - Q1
df = df[~((df['numeric_col'] < (Q1 - 1.5 * IQR)) | (df['numeric_col'] > (Q3 + 1.5 * IQR)))]

# 7. Standardize string formatting
df['text_col'] = df['text_col'].str.strip().str.lower()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Groupby aggregation — mean of numeric column per category
df.groupby('category_col')['numeric_col'].mean()

# Multiple aggregations at once
df.groupby('category_col')['numeric_col'].agg(['mean', 'median', 'std', 'count'])

# Skewness and kurtosis (shape of distribution)
df['numeric_col'].skew()
df['numeric_col'].kurt()

# Scatter plot between two numeric columns
sns.scatterplot(x='col1', y='col2', data=df)
plt.show()

# Violin plot (distribution + boxplot combined)
sns.violinplot(x='category_col', y='numeric_col', data=df)
plt.show()

# Countplot for categorical frequency
sns.countplot(x='category_col', data=df)
plt.show()

# Crosstab between two categorical columns
pd.crosstab(df['category_col1'], df['category_col2'])
```
